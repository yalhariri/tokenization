This notebook aims to simplify building a tokenization model in the Text processing and LLM modelling


```python
import sentencepiece as spm

## Train a SentencePiece model by using botchan.txt as input file (the larger the better)
spm.SentencePieceTrainer.train('--input=botchan.txt --model_prefix=trained_model --vocab_size=44 --model_type=bpe')

sp = spm.SentencePieceProcessor()
sp.load('trained_model.model')
```

```python
test_sentence = 'to see what he could see see see'
encoded_tokens = sp.tokenize(test_sentence, out_type=str)
encoded_ids = sp.tokenize(test_sentence, out_type=int)
print(f"Encoded tokens: {encoded_tokens}")
print(f"Encoded IDs: {encoded_ids}")


# decode: id => text

print(f"Decoded tokens: {sp.decode_pieces(encoded_tokens)}")
print(f"Decoded IDs: {sp.decode_ids(encoded_ids)}")

```
